```python
# 1. 安装魔搭库和指定版本的 transformers
!pip install modelscope transformers==4.39.3

from modelscope import snapshot_download
from transformers import AutoTokenizer, AutoModel

print("🚀 正在通过阿里云内网下载模型，速度会非常快...")
# 2. 从魔搭社区下载模型到本地缓存
model_dir = snapshot_download("ZhipuAI/chatglm3-6b", revision="v1.0.0")
print(f"✅ 模型下载完成！保存在: {model_dir}")

print("🧠 正在将模型加载到 GPU 显存中...")
# 3. 从本地目录加载模型
tokenizer = AutoTokenizer.from_pretrained(model_dir, trust_remote_code=True)
model = AutoModel.from_pretrained(model_dir, trust_remote_code=True, device='cuda')
model = model.eval()

print("🎉 模型加载成功！可以开始对话了。")
```

```python
# 初始化空的对话历史列表
history = []
print("🤖 ChatGLM3-6B 助手已启动！(输入 '退出' 或 'exit' 结束对话)")

while True:
    # 获取你的输入
    user_input = input("\n🧑‍💻 你: ")

    # 设置退出条件
    if user_input.strip().lower() in ['退出', 'exit', 'quit']:
        print("👋 对话已结束。")
        break

    # 防止空输入报错
    if not user_input.strip():
        continue

    # 调用模型推理：传入分词器、当前问题，以及历史上下文
    # 模型会返回生成的回复，并自动更新 history 列表
    response, history = model.chat(tokenizer, user_input, history=history)

    # 打印模型的回答
    print(f"\n🤖 ChatGLM: {response}")
    print("-" * 50) # 打印一条分割线，方便阅读
```
